```python
# Import library
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

Import the libraries needed for plotting and data analysis.

Create a simple line graph using four y-values.

```python
# Plot x versus y
plt.plot([1,2,3,4], [1,4,9,16])
```

```python
# Plot with red circles
plt.plot([1,2,3,4],[1,4,9,16],'ro')
plt.axis([0,6,0,20])
plt.show()
```

```python
# Evenly sampled time at 200ms intervals
t = np.arange(0., 5., 0.2)
t
```

Create evenly spaced numerical values from 0 to 5 in intervals of 0.2.

```python
# Red dashes, blue squares and green triangles
plt.plot(t, t, 'r--', t, t**2, 'bs', t, t*3, 'g^')
plt.show()
```

```python
#
data = {'a': np.arange(50),
        'c': np.random.randint(0,50,50),
        'd': np.random.randn(50)}
data
```

Create a sample dataset containing sequential and randomly generated values.

```python
data['b'] = data['a'] + 10 * np.random.randn(50)
data
```

Add a new variable, b, based on a plus random variation.

```python
data['d'] = np.abs(data['d']) * 100
data
```

Convert d to positive values and increase its scale.

```python
# Plot with keyword strings
plt.scatter('a', 'b', c='c', s='d', data=data)
plt.xlabel('entry a')
plt.ylabel('entry b')
plt.show()
```

Create a scatterplot where position, color, and point size represent different variables.

```python
# Plot with categorial variables
names = ['group_a',  'group_b', 'group_c']
values = [1, 10, 100]

plt.figure(figsize=(9,3))

plt.subplot(131)
plt.bar(names, values)
plt.subplot(132)
plt.scatter(names, values)
plt.subplot(133)
plt.plot(names, values)
plt.suptitle('Categorial Plotting')
plt.show()
```
